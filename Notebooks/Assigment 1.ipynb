```python
# normalise()

def normalise(values, minimum=None, maximum=None):

    if len(values) == 0:
        return []

    if minimum is None:
        minimum = min(values)

    if maximum is None:
        maximum = max(values)

    if minimum == maximum:
        return [0.0 for value in values]

    result = []

    for value in values:
        new_value = (value - minimum) / (maximum - minimum)
        result.append(new_value)

    return result
```

```python
print(normalise([13, 10, 28, 90]))

print(normalise([5, 8, 30, 9]))
```

```python
# summarise_scores()

def summarise_scores(scores):
    if len(scores) == 0:
        return {
            "count": 0,
            "mean": None,
            "minimum": None,
            "maximum": None,
            "above_threshold": 0
        }

    count = len(scores)
    mean = sum(scores) / count
    minimum = min(scores)
    maximum = max(scores)
    above_threshold = sum(score >= 0.8 for score in scores)

    return {
        "count": count,
        "mean": mean,
        "minimum": minimum,
        "maximum": maximum,
        "above_threshold": above_threshold
    }
```

```python
print(summarise_scores([0.6, 0.75, 0.85, 0.9]))

print(summarise_scores([0.5, 0.7, 0.82]))
```

```python
print(summarise_scores([]))
{'count': 0, 'mean': None, 'minimum': None, 'maximum': None, 'above_threshold': 0}
```

```python
# safe_divide()

def safe_divide(numerator, denominator, default=0.0):
    try:
        return numerator / denominator
    except (ZeroDivisionError, TypeError):
        return default
```

```python
print(safe_divide(18, 2))
print(safe_divide(10, 0))
print(safe_divide("10", 2))
```
